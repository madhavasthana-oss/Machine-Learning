```python
import sklearn as skl
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
Y = iris.target
```

```python
from sklearn.preprocessing import StandardScaler
X_transformed = StandardScaler().fit_transform(X=X)
```

```python
from sklearn.decomposition import PCA
X_pca = PCA(n_components=2).fit_transform(X_transformed)
```

```python
print('Shape before PCA: ', X_transformed.shape) 
print('Shape after PCA: ', X_pca.shape)
```

```python
import pandas as pd
pca_df = pd.DataFrame( 
    data=X_pca,  
    columns=[ 
        'Principal Component 1',  
        'Principal Component 2'
    ]
)
```

```python
target_names = {
    0:"setosa",
    1:"versicolor",
    2:"virginica"
}

pca_df["target"] = Y
pca_df["target"] = pca_df["target"].map(target_names)
pca_df.sample(10)
```

```python
import matplotlib.pyplot as plt  
import seaborn as sns 
sns.set() 
# Plot 2D PCA Graph 
sns.lmplot( 
    x='Principal Component 1',  
    y='Principal Component 2',  
    data=pca_df,  
    hue='target',  
    fit_reg=False,  
    legend=True 
    ) 
  
plt.title('2D PCA Graph of iris dataset')
plt.show()
```

```python
# Pedhazur formulas
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x ** 2)

B1_ped = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)
B0_ped = (sum_y - B1_ped * sum_x) / n

print("=== Pedhazur Method ===")
print(f"B0 (Intercept) = {B0_ped:.4f}")
print(f"B1 (Slope)     = {B1_ped:.4f}")
```

```python
# Solving the 2x2 system
A = np.array([
    [n, sum_x],
    [sum_x, sum_x2]
])
b = np.array([sum_y, sum_xy])

B0_calc, B1_calc = np.linalg.solve(A, b)

print("\n=== Calculus Method ===")
print(f"B0 (Intercept) = {B0_calc:.4f}")
print(f"B1 (Slope)     = {B1_calc:.4f}")
```

```python
# Using Pedhazur coefficients (they will be identical)
y_pred = B0_ped + B1_ped * x

# RMSE
rmse = np.sqrt(np.mean((y - y_pred) ** 2))

print("\n=== Results ===")
print(f"Predicted responses: {np.round(y_pred, 4)}")
print(f"RMSE = {rmse:.4f}")

print("\n=== Comparison ===")
print(f"Pedhazur  → B0 = {B0_ped:.6f}, B1 = {B1_ped:.6f}")
print(f"Calculus  → B0 = {B0_calc:.6f}, B1 = {B1_calc:.6f}")
print("Both methods give identical results (as expected).")
```
