```python
import numpy as np

def gradient_descent_regression(X, y, lr=0.0001, batch_size=1, num_epochs=30):
    n = len(y)
    beta = np.zeros(X.shape[1])
    losses = []
    for epoch in range(num_epochs):
        # Shuffle examples, keeping X and y aligned
        perms = np.random.permutation(n)
        X_epoch = X[perms]
        y_epoch = y[perms]

        print(f"<<< EPOCH -> (epoch) >>>")

        for i in range(0, n, batch_size):
            X_sec = X_epoch[i:i + batch_size]
            y_sec = y_epoch[i:i + batch_size]

            # Forward pass
            pred = X_sec @ beta

            # Prediction error
            err = pred - y_sec

            # Gradient of batch MSE
            m = len(y_sec)
            grads = (2 / m) * X_sec.T @ err

            # Update coefficients
            beta -= lr * grads

            # track losses
            loss = np.mean((X @ beta - y)**2)
            losses.append(loss)
    print("<<< finished >>>")
    return beta, losses
```

```python
def sigmoid(z):
    return (1 / 1 + np.exp(
        np.clip(z, -500, 500)
    ))

def gradient_descent_logistic(
    X,
    y,
    lr = 0.001,
    num_epochs = 10,
    batch_size = 1):
    x = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float).reshape(-1)
    n = len(y)
    losses = []
    beta = np.zeros(X.shape[1])
    for epoch in range(num_epochs):
        perms = np.random.permutation(n)
        x, y  = x[perms], y[perms]
        for i in range(0, n, batch_size):
            xi = x[i:i+batch_size]
            yi = y[i:i+batch_size]

            # get predictions and error
            preds = sigmoid(xi @ beta)
            err = preds - yi
            
            # gradients
            grads = xi.T.dot(err).flatten()
            beta  = beta - lr * grads 

            # loss
            p = sigmoid(x.T.dot(beta))
            loss = -np.mean(
                y * np.log(p + 1e-15) + (1 - y) * np.log(1 - p + 1e-15)
            )
            
            losses.append(loss)
    return beta, losses




```

```python
import numpy as np
def get_distance(p1, p2):
    
    return np.linalg.norm(np.asarray(p1) - np.asarray(p2))
```

```python
points = [
    (15,39), (16,81), (17,6), (19,77), (25,40), (26,70), (30,88), (35,29), (45,85), (55,52)
]

def get_dist_df(points, k:int = 1):
    dists = {}
    if k == 1:
        c = [(30, 88)]
    if k == 2:
        c = [(25,40), (45,85)]
    if k == 3:
        c = [(16,81), (19,77), (55,52)]
    if k == 4:
        c = [(15,39), (17,6), (26,70), (35,29)]
    for i,c_ in enumerate(c):    
        dists[f"c_{i}"] = [
            get_distance(x, c_) for x in points 
        ]
    return pd.DataFrame(dists)
```

```python
def assign_clusters(dist_df):
    return dist_df.values.argmin(axis=1)
```

```python
def kmeans(points, k, n_iter = 100):
    points = np.array(points)
    for i in range(n_iter):
        df = get_dist_df(points, k = k)
        labels = assign_clusters(df)

        centroids = [
            points[labels == j].mean(axis = 0).tolist()
            for j in range(k)
        ]
    
    return labels, centroids
```
