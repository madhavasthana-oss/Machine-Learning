```python
# question 3
x = np.array([21, 64, 86, 22, 74, 55, 81, 79, 90, 89])
y = np.array([21, 7, 3, 45, 10, 29, 55, 4, 37, 18])
```

```python
x_out_greater = np.array([])
x_out_equal   = np.array([])
```

```python
for idx, n in enumerate(x):
    if n == y[idx]:
        x_out_equal = np.append(
            x_out_equal, idx
        )
    elif n > y[idx]:
        x_out_greater = np.append(
            x_out_greater,idx
        ) 
    else:
        continue
```

```python
x_out_greater
```

```python
x_out_equal
```
